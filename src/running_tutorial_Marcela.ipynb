# 1. Running the model
    
Inside your terminal go to the folder src/ 
    
    

### 1.a. Run the follow command to clear your python cache and deletes some compiled files

``make clean``

### 1.b. Compile fortran source and creates a file named with something like `caete_module.so`, the wrapped Fortran functions and subroutines for python.


``make so`` 

#### 1.c. Determines the number of initial PLSs [TO BE CHANGED]
``Number of initial random PLS:``

In general, we use ~3000 PLSs. After a sensitivity test I found out that a good number for the allometry version is 6000. 

For the moment please use 6000. We can adapt it after. I will explain where you can change and why you shouldn't change it now.


### 1.d. Run the model

`python model_driver.py`

#### 1.d.1. Choose the spatial scale: 
        
``THREE MASK OPTIONS: AMAZON BIOME (a); PAN-AMAZON (b) OR PLOT RUN (c)``

Here you choose which mask you will use. This mask determines from which gridcells you will use the input files. 
I suggest using the mask (a) and then you decide which gridcells will be run after. Hence, you ensure all the gridcells are correctly initialized
    


#### 1.d.2. Choose the model version

``Which version? (1: allom/2: nutri_cycle)``

-Type 1 for the version using allometric constraints to determine allocation. Also, in this version some traits are put as fixed (see it in plsgen.py)

-Type 2 for the version that considers nutrients cycle and the allocation is determined by fractions of NPP as variant functional traits. 

#### 1.d.2. Inform if you are running in Sombrero server or in your own computer

``RUN IN SOMBRERO(y/n)``

I strongly suggest you answer "n" because I changed some paths. There is no problem here if you say "no" because it just configure some output path

#### 1.d.3. Nominate the folder for your outputs (ATTENTION: make sure the name have no spaces)

``Set the folder to store outputs:
Give a name to your run (ASCII letters and numbers only. No spaces):``

I suggest choosing a name that describe the running your are doing. Plus, take notes about this running in some file or notebook

OBS: Note that the path in which the files are saved are indicated in the terminal. For example:

``The raw model results & the PLS table will be saved at: /d/c1/homes/amazonfaceme/biancarius/CAETE-DVM-alloc-allom/outputs/example_Marcela``

``The final netCDF files will be stored at: /d/c1/homes/amazonfaceme/biancarius/CAETE-DVM-alloc-allom/outputs/example_Marcela/nc_outputs``

#### 1.d.4. Choose the specific region you want to simulate:
                
``Select a zone [c: central, s: south, e: east, nw: NW]``

**NOTE**: You can change the grid cells/region to be simulated in the file model_driver.py between the lines 151 and 169. 
You can choose just one gridcell or a region. 

Between the lines 151 and 169 you define y0, y1, x0 and x1. This notation is like you are choosing a specific location in the map. The 'y' determines the latitude and the 'x' the longitude. If you have the lat/long in decimals you can use the file "find_coordinates.py" in the folder src/ to find the correspondent values for CAETÊ coordinates.

An example, if you want to run a unique gridcell that has the CAETÊ coordinates y = 186 and x = 239, you need to put your y0 as 186 and your y1 as y0+1, and your x0 as 239 and your x1 as x0+1. Because of how python works you can't use only 186 and 239. See: 

``Gridcell MAN: 186, 239``

``y0, y1 = 186, 187``

``x0, x1 = 239, 240``

But if you want to run a region, you need to determine the interval of y and x.


#### 1.d.5. Choose the gridcell acronym [TO BE CHANGED]

``Gridcell acronym [AFL, ALP, FEC, MAN, CAX]:``

For the moment, type MAN. If you want to run the model in another place that is not Manaus, please let me know and I'll change it. The change need to be done in the file parameters.py

# IMPORTANT INFORMATION:

The model will be run from 1979 to 2016. You can change this period in the file caete.py.
Don't forget to verify if the inputs match with the period of time you want to run.
